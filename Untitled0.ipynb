## 🎤 都道府県テーマ ラップ生成テンプレート 🎤

このテンプレートは、日本の特定の都道府県をテーマにしたラップソングを作成するのに役立ちます。各セクションの指示に従って、都道府県のユニークな特徴、文化、スポット、感情を表現しましょう。

### テンプレートの構成要素:

1.  **Intro (イントロ)**:
    *   都道府県のキャッチフレーズ、またはその都道府県に対する一般的な認識（良い点、悪い点、皮肉など）。
    *   その都道府県を象徴する具体的な場所やアイテムを提示。

2.  **Verse 1 (ヴァース1)**:
    *   都道府県の日常風景、有名な場所、食べ物、特産品などを具体的に描写。
    *   交通手段（電車、道路など）や地理的特徴（山、川、海など）に触れる。
    *   住民の行動パターンや独特の習慣を表現する。

3.  **Hook (フック)**:
    *   曲のテーマとなる繰り返しの部分。都道府県の魂、主要なメッセージ、または最も印象的な特徴を凝縮する。
    *   記憶に残りやすいフレーズや、その都道府県を代表する言葉を盛り込む。

4.  **Verse 2 (ヴァース2)**:
    *   都道府県の歴史、文化、祭り、イベントなどに焦点を当てる。
    *   有名人、スポーツチーム、企業など、その都道府県を象徴する要素。
    *   他の都道府県との比較や、住民が感じる誇りや課題を表現する。

5.  **Outro (アウトロ)**:
    *   曲の締めくくり。都道府県への愛着、未来への展望、またはリスナーへのメッセージ。
    *   イントロで触れたテーマへの言及や、感動的な一言で締めくくる。

### 作成のヒント:

*   **具体的な固有名詞**: 地名、駅名、お店の名前、特産品名などを積極的に使うとリアリティが増します。
*   **五感に訴える**: 見えるもの、聞こえるもの、匂い、味、感触などを描写して情景を豊かにします。
*   **比喩やパンチライン**: 独自性のある表現や、聞く人をハッとさせるような言葉を盛り込みましょう。
*   **韻を踏む**: ラップの基本である韻を意識して言葉を選ぶと、リズム感が生まれます。
*   **方言や口癖**: その都道府県ならではの言葉を使うと、より地域性が際立ちます。

次に、このテンプレートを使ってラップを生成するためのPython関数を提供します。


```python
def generate_prefecture_rap(
    prefecture_name,
    intro_catchphrase,
    intro_symbol,
    verse1_details,
    verse1_transport,
    verse1_culture,
    hook_main_message,
    hook_slogan,
    verse2_history,
    verse2_icons,
    verse2_challenges,
    outro_final_words
):
    """
    特定の都道府県をテーマにしたラップを生成するテンプレート関数。

    Args:
        prefecture_name (str): 都道府県の名前（例: '埼玉県'）。
        intro_catchphrase (str): イントロのキャッチフレーズや一般的な認識。
        intro_symbol (str): イントロで提示する具体的な象徴。
        verse1_details (list): ヴァース1で描写する日常風景、場所、食べ物などのリスト。
        verse1_transport (str): ヴァース1で触れる交通手段や地理。
        verse1_culture (str): ヴァース1で触れる住民の習慣や文化。
        hook_main_message (str): フックの主要なメッセージ。
        hook_slogan (str): フックの記憶に残るスローガン。
        verse2_history (str): ヴァース2で触れる歴史やイベント。
        verse2_icons (list): ヴァース2で触れる有名人、チーム、企業などのリスト。
        verse2_challenges (str): ヴァース2で触れる誇りや課題。
        outro_final_words (str): アウトロの締めくくりの言葉。

    Returns:
        str: 生成されたラップソングのテキスト。
    """

    rap_lyrics = f"""
🎧「{prefecture_name}フロウ」🎧

[Intro]
{intro_catchphrase}
{intro_symbol}

[Verse 1]
{chr(10).join(verse1_details)}
{verse1_transport}
{verse1_culture}
それが{prefecture_name}のスタイル

[Hook]
{hook_main_message}
{hook_slogan}
{prefecture_name}ソウルは
街のどこかに息づいてる

[Verse 2]
{verse2_history}
{chr(10).join(verse2_icons)}
{verse2_challenges}
それが俺らのプライド

[Outro]
{outro_final_words}


"""
    return rap_lyrics

print("ラップ生成テンプレート関数 'generate_prefecture_rap' が定義されました。")
print("引数に値を渡して、ラップを生成してみましょう！")
```

```python
from moviepy.editor import ImageClip, AudioFileClip, concatenate_videoclips
import os

# サンプル画像ファイルと音声ファイルを用意します。
# 実際のパスに置き換えてください。

# 新しいダミー画像ファイルを作成（異なる色と名前を使用）
from PIL import Image
import numpy as np

def create_dummy_image(filename, color, size=(640, 480)):
    img = Image.new('RGB', size, color=color)
    img.save(filename)

create_dummy_image('image_new1.png', 'orange')
create_dummy_image('image_new2.png', 'purple')
create_dummy_image('image_new3.png', 'yellow')

# サンプル音声ファイルをダウンロード（まだダウンロードしていない場合）
# コマンドをコメント解除して実行してください。
# !wget -O audio.mp3 'https://www.soundhelix.com/examples/mp3/SoundHelix-Song-3.mp3'

image_files = ['image_new1.png', 'image_new2.png', 'image_new3.png']
audio_file = 'audio.mp3' # 実際の音声ファイルパスに置き換えてください
output_video_file = 'output_video_new.mp4' # 出力ファイル名も変更

# ダミーの音声ファイルがなければ警告を出す
if not os.path.exists(audio_file):
    print(f"警告: 音声ファイル '{audio_file}' が見つかりません。有効な音声ファイルをダウンロードするか、パスを修正してください。")
    print("動画に音声は含まれません。\n例: !wget -O audio.mp3 'https://www.soundhelix.com/examples/mp3/SoundHelix-Song-3.mp3'")
    audio_clip = None
else:
    audio_clip = AudioFileClip(audio_file)

videoclips = []

# 各画像ファイルをビデオクリップに変換
for i, img_file in enumerate(image_files):
    # 画像を表示する秒数を設定
    duration_per_image = 2 # 各画像を2秒表示
    clip = ImageClip(img_file, duration=duration_per_image)
    videoclips.append(clip)

# すべての画像クリップを結合
final_clip = concatenate_videoclips(videoclips)

# 音声クリップを動画に設定
if audio_clip:
    # 動画の長さに合わせて音声クリップをトリミングまたはループ
    if audio_clip.duration < final_clip.duration:
        # 音声が短い場合、ループさせる
        audio_clip = audio_clip.audio_loop(duration=final_clip.duration)
    elif audio_clip.duration > final_clip.duration:
        # 音声が長い場合、動画の長さにトリミング
        audio_clip = audio_clip.subclip(0, final_clip.duration)
    final_clip = final_clip.set_audio(audio_clip)

# 動画を書き出し
final_clip.write_videofile(output_video_file, fps=24, codec="libx264")

print(f"動画 '{output_video_file}' が作成されました。")

# ダミーファイルを削除 (オプション)
os.remove('image_new1.png')
os.remove('image_new2.png')
os.remove('image_new3.png')
# if os.path.exists(audio_file): os.remove(audio_file) # 実際にダウンロードした場合
```

Next, I'll define helper functions to create images with text and to convert text to speech. Then, I'll use these to create a video for the '浦和のうなこちゃん' rap.

```python
from PIL import Image, ImageDraw, ImageFont
from gtts import gTTS
from moviepy.editor import ImageClip, AudioFileClip, concatenate_videoclips, TextClip, CompositeVideoClip
import os

# フォントの準備（Colabに存在するフォントを使用）
# /usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf など
# 日本語フォントが必要な場合は、別途インストールまたはダウンロードしてください。
# 例: !apt-get update && apt-get install -y fonts-ipafont-gothic
font_path = '/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf' # デフォルトのフォント
# 日本語フォントをインストールしている場合は以下を使用
# font_path = '/usr/share/fonts/truetype/fonts-japanese-gothic.ttf'
try:
    # 実際の日本語フォントのパスに調整してください
    japanese_font_path = '/usr/share/fonts/truetype/noto/NotoSansCJK-Regular.ttc'
    # または他の日本語フォントのパス
    ImageFont.truetype(japanese_font_path, 40)
    font_path = japanese_font_path
except IOError:
    print(f"Warning: Japanese font not found at {japanese_font_path}. Using default font. Japanese characters might not display correctly.")

def create_text_image(text, output_filename, img_size=(1280, 720), bg_color=(0, 0, 0), text_color=(255, 255, 255), font_size=40):
    img = Image.new('RGB', img_size, color = bg_color)
    d = ImageDraw.Draw(img)

    try:
        font = ImageFont.truetype(font_path, font_size)
    except IOError:
        font = ImageFont.load_default() # Fallback to default font
        print("Warning: Custom font not found, using default font.")

    # テキストを改行処理して中央揃え
    lines = []
    current_line = []
    words = text.split()
    for word in words:
        test_line = ' '.join(current_line + [word])
        bbox = d.textbbox((0, 0), test_line, font=font)
        text_width = bbox[2] - bbox[0]
        if text_width < img_size[0] - 40: # 40px margin
            current_line.append(word)
        else:
            lines.append(' '.join(current_line))
            current_line = [word]
    lines.append(' '.join(current_line))

    # すべての行の高さを計算
    total_text_height = sum([d.textbbox((0, 0), line, font=font)[3] - d.textbbox((0, 0), line, font=font)[1] for line in lines])
    total_text_height += (len(lines) - 1) * (font_size // 5) # 行間

    y_text = (img_size[1] - total_text_height) / 2

    for line in lines:
        bbox = d.textbbox((0, 0), line, font=font)
        line_width = bbox[2] - bbox[0]
        x_text = (img_size[0] - line_width) / 2
        d.text((x_text, y_text), line, font=font, fill=text_color)
        y_text += (bbox[3] - bbox[1]) + (font_size // 5) # 次の行のy座標を更新（行間を考慮）

    img.save(output_filename)
    return output_filename

def text_to_speech(text, output_audio_file, lang='ja'):
    tts = gTTS(text=text, lang=lang, slow=False)
    tts.save(output_audio_file)
    return output_audio_file

print("画像と音声生成のヘルパー関数が定義されました。")
```

```python
# 浦和のうなこちゃんラップの動画を作成

rap_text = unakochan_rap
rap_title = "浦和のうなこちゃんフロウ - ミュージックビデオ"
output_video_file_unakochan = 'unakochan_rap_video.mp4'

# 音声ファイル生成
output_audio_file_unakochan = 'unakochan_rap_audio.mp3'
text_to_speech(rap_text, output_audio_file_unakochan)

# ラップのセクションを分割
sections = {
    "Title": rap_title,
    "Intro": "",
    "Verse 1": "",
    "Hook": "",
    "Verse 2": "",
    "Outro": ""
}

current_section = ""
for line in rap_text.splitlines():
    line = line.strip()
    if line.startswith('🎧'):
        continue
    elif line == '[Intro]':
        current_section = "Intro"
    elif line == '[Verse 1]':
        current_section = "Verse 1"
    elif line == '[Hook]':
        current_section = "Hook"
    elif line == '[Verse 2]':
        current_section = "Verse 2"
    elif line == '[Outro]':
        current_section = "Outro"
    elif current_section and line:
        sections[current_section] += line + "\n"

# タイトル画像を生成
title_image_file = create_text_image(rap_title, 'unakochan_title.png', font_size=60)

# 各セクションの画像を生成
image_files = [title_image_file]
for section_name, section_text in sections.items():
    if section_name != "Title" and section_text.strip():
        image_filename = f'unakochan_{section_name.lower().replace(" ", "_")}_lyrics.png'
        create_text_image(section_text.strip(), image_filename)
        image_files.append(image_filename)

# MoviePyで動画を結合
clips = []
# タイトルクリップは少し長めに表示
clips.append(ImageClip(title_image_file).set_duration(3))

# 各歌詞クリップの表示時間を計算（音声の長さに合わせるか、固定時間）
# ここでは、簡略化のため、全体の音声の長さを画像の数で均等に割り振ります。
# より高度な同期には、各セクションの音声の長さを個別に測定する必要があります。

audio_clip_full = AudioFileClip(output_audio_file_unakochan)

# タイトルを除く歌詞画像の数
num_lyric_images = len(image_files) - 1
# 各歌詞画像の表示時間 (タイトル時間は除く)
if num_lyric_images > 0:
    duration_per_lyric_image = (audio_clip_full.duration - 3) / num_lyric_images
else:
    duration_per_lyric_image = 2 # fallback if no lyrics

current_time = 3 # タイトル画像の表示時間から開始
for img_file in image_files[1:]:
    clip = ImageClip(img_file).set_duration(duration_per_lyric_image)
    clips.append(clip)

final_clip_images = concatenate_videoclips(clips)
final_video_clip = final_clip_images.set_audio(audio_clip_full)

final_video_clip.write_videofile(output_video_file_unakochan, fps=24, codec="libx264")

print(f"動画 '{output_video_file_unakochan}' が作成されました。")

# 生成された一時ファイルをクリーンアップ
os.remove(output_audio_file_unakochan)
for img_file in image_files:
    os.remove(img_file)
```

```python
from PIL import Image, ImageDraw, ImageFont
from gtts import gTTS
from moviepy.editor import ImageClip, AudioFileClip, concatenate_videoclips, TextClip, CompositeVideoClip
import os

# 日本語フォントのパスを更新
font_path = '/usr/share/fonts/truetype/fonts-japanese-gothic.ttf'
try:
    ImageFont.truetype(font_path, 40)
    print(f"Using Japanese font at {font_path}")
except IOError:
    print(f"Error: Japanese font not found at {font_path}. Please check the installation.")
    font_path = '/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf' # Fallback
    print("Falling back to default font. Japanese characters might not display correctly.")

def create_text_image(text, output_filename, img_size=(1280, 720), bg_color=(0, 0, 0), text_color=(255, 255, 255), font_size=40):
    img = Image.new('RGB', img_size, color = bg_color)
    d = ImageDraw.Draw(img)

    try:
        font = ImageFont.truetype(font_path, font_size)
    except IOError:
        font = ImageFont.load_default() # Fallback to default font
        print("Warning: Custom font not found, using default font.")

    # テキストを改行処理して中央揃え
    lines = []
    current_line = []
    words = text.split()
    for word in words:
        test_line = ' '.join(current_line + [word])
        # bbox = d.textbbox((0, 0), test_line, font=font)
        # text_width = bbox[2] - bbox[0]
        # PILのtextbboxが単語分割と相性が悪い場合があるので、簡略化して文字数で制御
        # またはより堅牢なテキストレイアウトライブラリを使用することも検討
        if font.getlength(test_line) < img_size[0] - 80: # 80px margin
            current_line.append(word)
        else:
            lines.append(' '.join(current_line))
            current_line = [word]
    lines.append(' '.join(current_line))

    # すべての行の高さを計算
    # bbox = [left, top, right, bottom] -> height = bottom - top
    total_text_height = 0
    for line_text in lines:
        if line_text:
            bbox = d.textbbox((0, 0), line_text, font=font)
            total_text_height += (bbox[3] - bbox[1])
    total_text_height += (len(lines) - 1) * (font_size // 5) # 行間

    y_text = (img_size[1] - total_text_height) / 2

    for line_text in lines:
        if line_text:
            bbox = d.textbbox((0, 0), line_text, font=font)
            line_width = bbox[2] - bbox[0]
            x_text = (img_size[0] - line_width) / 2
            d.text((x_text, y_text), line_text, font=font, fill=text_color)
            y_text += (bbox[3] - bbox[1]) + (font_size // 5) # 次の行のy座標を更新（行間を考慮）

    img.save(output_filename)
    return output_filename

def text_to_speech(text, output_audio_file, lang='ja'):
    tts = gTTS(text=text, lang=lang, slow=False)
    tts.save(output_audio_file)
    return output_audio_file

print("画像と音声生成のヘルパー関数が日本語フォント対応で再定義されました。")
```

```python
import os
import pandas as pd

# Google DriveのMy Rapsディレクトリを基準にファイル一覧を取得
root_dir = '/content/drive/MyDrive/My Raps'  # My Rapsディレクトリ

file_data = []

for dirpath, dirnames, filenames in os.walk(root_dir):
    for f in filenames:
        filepath = os.path.join(dirpath, f)
        file_data.append({'Path': filepath, 'Type': 'File'})
    for d in dirnames:
        dirpath_full = os.path.join(dirpath, d)
        file_data.append({'Path': dirpath_full, 'Type': 'Directory'})

# DataFrameを作成
df_files = pd.DataFrame(file_data)

# CSVファイルとして保存
output_csv_path = 'my_raps_file_list.csv' # 新しいCSVファイル名
df_files.to_csv(output_csv_path, index=False, encoding='utf-8')

print(f"Google Driveの'My Raps'ディレクトリ内のファイルとディレクトリの一覧を '{output_csv_path}' に保存しました。")
display(df_files.head())
```
